# PyTorch Training Pipeline

```python
import torch
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
df.head()
```

Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

Scaling (uniform data)

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

Label encoding (for target labels)

```python
y_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

Converting NumPy arrays to PyTorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor, X_train_tensor.shape
```

# Defining the Model

note: OOP is gon be used a lot while constructing NNs

```python
class NeuralNetwork:
  def __init__(self, X):
    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True) #X.shape[1] gives the no of features corresponding to which we initialize the weights
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True) # a single bias corresponding to the single neuron

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # clamp prediction to avoid log(0)
    epsilion = 1e-7
    y_pred = torch.clamp(y_pred, epsilion, 1 - epsilion)

    # calculate loss (binary cross-entropy loss)
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

Hyperparameters

```python
learning_rate = 0.1
epochs = 100
```

# Training Pipeline

```python
# 1. create the model
model = NeuralNetwork(X_train_tensor)

# 2. Define loop
for epoch in range(epochs):

  # 2.2 forward pass
  y_pred = model.forward(X_train_tensor)

  # 2.3 loss calculation
  loss = model.loss_function(y_pred, y_train_tensor)

  # 2.4 backward propogation
  loss.backward()

  # 2.5. update parameters (gradient descent)
  with torch.no_grad(): # imp: we wanna perform these two operations without being tracked by the autograd (computational graph)
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # to prevent piliong up of gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()
  if not (epoch + 1) % 10:
    print(f'Epoch: {epoch + 1} -> Loss: {loss.item()}')
```

```python
model.weights, model.bias
```

Evaluation

```python
with torch.no_grad():
  # test on test data
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float() # 0 / 1

  # compute accuracy
  accuracy = (y_pred == y_test_tensor).float().mean()
print(f'Accuracy: {accuracy.item()}')

# print(y_pred)
```
